# Build Lublin City Boundary

This notebook creates the administrative boundary of Lublin from the official district boundary dataset.

The workflow:
- defines the project paths,
- locates the district shapefile,
- loads and inspects the district geometries,
- dissolves all district polygons into a single city boundary,
- saves the resulting boundary for use as the Area of Interest (AOI) in the subsequent analyses.

## Project paths and libraries

GeoPandas is used to process the vector boundaries. Project paths are defined relative to the repository root so that the notebook can be executed without hard-coded local file paths.

```python
from pathlib import Path
import geopandas as gpd

PROJECT_ROOT = Path.cwd().parent
DISTRICTS_DIR = PROJECT_ROOT / "data" / "boundaries" / "districts"
CITY_DIR = PROJECT_ROOT / "data" / "boundaries" / "city"

CITY_DIR.mkdir(parents=True, exist_ok=True)

print("Districts:", DISTRICTS_DIR)
print("City:", CITY_DIR)
```

## Locate the district boundary dataset

The available shapefile in the district boundary directory is listed as a quick input-data check.

```python
shp_files = list(DISTRICTS_DIR.glob("*.shp"))

print("Shapefiles found:")
for shp in shp_files:
    print(shp.name)
```

## Load and inspect district boundaries

The official district boundary dataset is loaded and checked for the number of features, coordinate reference system and basic attribute structure.

The dataset should contain 27 district polygons.

```python
districts_path = DISTRICTS_DIR / "dzielnice_granice.shp"

districts = gpd.read_file(districts_path)

print("Number of features:", len(districts))
print("CRS:", districts.crs)

display(districts.head())
```

## Create the city boundary

All district polygons are dissolved into a single geometry representing the administrative boundary of Lublin.

The original coordinate reference system is preserved during this operation.

```python
city = districts.dissolve()

print("Number of features after dissolve:", len(city))
print("Geometry type:", city.geometry.iloc[0].geom_type)
```

## Save the city boundary

The dissolved city geometry is saved as a shapefile in the project data directory.

This boundary is used as the common Area of Interest (AOI) for the Landsat analyses in the following notebooks.

```python
city_output = CITY_DIR / "lublin_city_boundary.shp"

city.to_file(city_output)

print("Saved:")
print(city_output)
```

## Output

The final city boundary is saved as:

`data/boundaries/city/lublin_city_boundary.shp`

This file is used as the Area of Interest in the Landsat preprocessing, classification, change detection and district-level analysis workflows.
